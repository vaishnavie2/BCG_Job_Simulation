## Overall Insights from EDA

• Most customers did not churn, while a smaller percentage churned, indicating class imbalance.

• Customers with lower energy consumption showed higher churn tendency, suggesting lower engagement customers are more likely to leave.

• Customers with lower profit margins were more likely to churn, indicating profitability is related to customer retention.

• Sales channel showed different churn patterns, indicating customer acquisition source may influence churn behavior.

• Overall, customer consumption behavior, profitability, and sales channel were important factors influencing churn.
